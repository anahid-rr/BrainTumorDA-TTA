```python
import os
import zipfile
import nibabel as nib
import numpy as np
from PIL import Image

# -----------------------------
# Paths
# -----------------------------
zip_file = "BraTS2019.zip"  # change to the uploaded zip file name
extract_dir = "BraTS_extracted2019"  # folder to extract zip
save_dir = "Flair_JPGs2019"          # folder to save converted jpgs

# Make directories if they don't exist
os.makedirs(extract_dir, exist_ok=True)
os.makedirs(save_dir, exist_ok=True)

# -----------------------------
# Unzip the dataset
# -----------------------------
with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"Dataset extracted to {extract_dir}")

# -----------------------------
# Convert .nii flair images to .jpg
# -----------------------------
for root, dirs, files in os.walk(extract_dir):
    for file in files:
        if "flair" in file.lower() and file.endswith(".nii") or file.endswith(".nii.gz"):
            nii_path = os.path.join(root, file)
            try:
                # Load NIfTI file
                img = nib.load(nii_path)
                data = img.get_fdata()

                # Take middle slice in axial plane
                slice_idx = data.shape[2] // 2
                slice_data = data[:, :, slice_idx]

                # Normalize to 0-255
                slice_norm = ((slice_data - np.min(slice_data)) / (np.ptp(slice_data)) * 255).astype(np.uint8)

                # Save as jpg
                base_name = os.path.splitext(os.path.basename(file))[0]
                jpg_path = os.path.join(save_dir, f"{base_name}.jpg")
                Image.fromarray(slice_norm).save(jpg_path)
            except Exception as e:
                print(f"Failed to process {file}: {e}")

print(f"Conversion complete. Flair JPGs saved in {save_dir}")
```
